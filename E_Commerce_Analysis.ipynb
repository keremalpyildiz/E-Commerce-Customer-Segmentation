```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans

url = "https://raw.githubusercontent.com/SteffiPeTaffy/machineLearningAZ/master/Machine%20Learning%20A-Z%20Template%20Folder/Part%204%20-%20Clustering/Section%2024%20-%20K-Means%20Clustering/Mall_Customers.csv"
df = pd.read_csv(url)

print("Dataset Dimensions (Rows, Columns):", df.shape)
print("-" * 30)
df.info()

plt.figure(figsize=(10, 6))
sns.scatterplot(x='Annual Income (k$)', y='Spending Score (1-100)', data=df, s=60, color='darkred')
plt.title('Customer Income and Spending Distribution (Raw Data)')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()

X = df.iloc[:, [3, 4]].values
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
df['Cluster'] = kmeans.fit_predict(X)

df.head()
```

```python
# 1. Setting the figure size for our final plot
plt.figure(figsize=(10, 6))

# 2. Creating the scatter plot, coloring the points based on their 'Cluster'
sns.scatterplot(x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster', data=df, palette='tab10', s=70, legend='full')

# 3. Adding title and axis labels
plt.title('Customer Segments after AI (K-Means Clustering)')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')

# 4. Displaying the plot
plt.show()
```
